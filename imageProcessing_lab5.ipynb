```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

%matplotlib inline
```

```python
image = data.astronaut()
sigma = 2
amount = 1.5

image_float = image.astype(np.float32) / 255.0

blurred = cv2.GaussianBlur(image_float, (0, 0), sigmaX=sigma, sigmaY=sigma)

sharpened = np.clip(image_float + amount * (image_float - blurred), 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image);      ax[0].set_title("Original Image");              ax[0].axis("off")
ax[1].imshow(blurred);    ax[1].set_title("Blurred Image");               ax[1].axis("off")
ax[2].imshow(sharpened);  ax[2].set_title("Sharpened (Unsharp Masking)"); ax[2].axis("off")
plt.show()
```

```python
image = data.astronaut()

image_float = image.astype(np.float32) / 255.0

kernel = np.ones((7, 7), dtype=np.float32) / 49.0

filtered = cv2.filter2D(image_float, -1, kernel)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(filtered)
ax[1].set_title("7x7 Box Filter")
ax[1].axis("off")

plt.show()
```

```python
image = data.astronaut()

image_float = image.astype(np.float32) / 255.0

gaussian_5x5 = cv2.GaussianBlur(image_float, (5, 5), 0)

gaussian_21x21 = cv2.GaussianBlur(image_float, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(gaussian_5x5)
ax[1].set_title("5x5 Gaussian Filter")
ax[1].axis("off")

ax[2].imshow(gaussian_21x21)
ax[2].set_title("21x21 Gaussian Filter")
ax[2].axis("off")

plt.show()
```

```python
image = data.astronaut()

image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

sharpened = np.clip(image_float - laplacian, 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(laplacian)
ax[1].set_title("Laplacian")
ax[1].axis("off")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")

plt.show()
```

```python
image = data.astronaut()

image_float = image.astype(np.float32) / 255.0

box_3x3 = cv2.blur(image_float, (3, 3))
box_9x9 = cv2.blur(image_float, (9, 9))
box_15x15 = cv2.blur(image_float, (15, 15))

fig, ax = plt.subplots(1, 4, figsize=(20, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(box_3x3)
ax[1].set_title("3x3 Box Filter")
ax[1].axis("off")

ax[2].imshow(box_9x9)
ax[2].set_title("9x9 Box Filter")
ax[2].axis("off")

ax[3].imshow(box_15x15)
ax[3].set_title("15x15 Box Filter")
ax[3].axis("off")

plt.show()
```

```python
image = data.astronaut()

image_uint8 = image.astype(np.uint8)

noisy = image_uint8.copy()

salt = np.random.random(image_uint8.shape[:2]) < 0.02
pepper = np.random.random(image_uint8.shape[:2]) < 0.02

noisy[salt] = 255
noisy[pepper] = 0

median = cv2.medianBlur(noisy, 5)

gaussian = cv2.GaussianBlur(noisy, (5, 5), 0)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

ax[0].imshow(noisy)
ax[0].set_title("Noisy Image")
ax[0].axis("off")

ax[1].imshow(median)
ax[1].set_title("Median Filter")
ax[1].axis("off")

ax[2].imshow(gaussian)
ax[2].set_title("Gaussian Blur")
ax[2].axis("off")

plt.show()
```

```python
image = data.astronaut()

bilateral = cv2.bilateralFilter(image, 9, 75, 75)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")

ax[1].imshow(bilateral)
ax[1].set_title("Bilateral Filter")
ax[1].axis("off")

plt.show()
```
